### Feature Comparisons

```python
sex_0_df = df[df['sex'] == 0]
sex_0_df = df[df['sex'] == 0].drop(columns=['sex', 'time_months'])
sex_1_df = df[df['sex'] == 1]
sex_1_df = df[df['sex'] == 1].drop(columns=['sex', 'time_months'])
```

```python
corr_matrix = sex_0_df.corr()
matrix = corr_matrix['outcome'].drop('outcome')
print(matrix)
```

```python
mask = np.triu(np.ones_like(corr_matrix, dtype=bool), k=1)

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix,
    mask=mask,
    cmap='coolwarm',
    vmax=1.0, vmin=-1.0,
    annot=True,
    fmt=".3f",
    square=True,
    linewidths=0.5
)
plt.xticks(rotation=45, ha='right')
plt.title('Female Correlation Matrix', fontsize=14, pad=15)
plt.show()
```

```python
corr_matrix = sex_1_df.corr()
matrix = corr_matrix['outcome'].drop('outcome')
print(matrix)
```

```python
mask = np.triu(np.ones_like(corr_matrix, dtype=bool), k=1)
plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix, 
    mask=mask,
    cmap='coolwarm',
    vmax=1.0, vmin=-1.0,
    annot=True,
    fmt=".3f",
    square=True,
    linewidths=0.5
)
plt.xticks(rotation=45, ha='right')
plt.title('Male Correlation Matrix', fontsize=14, pad=15)
plt.show()
```

```python
df2 = df.drop(columns=['time_months'])
corr_matrix = df2.corr()
matrix = corr_matrix['outcome'].drop('outcome')
print(matrix)
```

```python
mask = np.triu(np.ones_like(corr_matrix, dtype=bool), k=1)

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr_matrix, 
    mask=mask,
    cmap='coolwarm',
    vmax=1.0, vmin=-1.0,
    annot=True,
    fmt=".3f",
    square=True,
    linewidths=0.5
)
plt.xticks(rotation=45, ha='right')
plt.title('Overall Correlation Matrix', fontsize=14, pad=15)
plt.show()
```
